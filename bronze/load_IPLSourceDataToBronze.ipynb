# Importing Libraries

```python
from pyspark.sql.functions import current_timestamp,col,input_file_name
```

# Read all JSON files

```python
readJsonFiles = spark.read.format("json").option('multiLine',True).load("/Volumes/workspace/source/sourceipldata/*.json")
```

# Adding extra columns such as Ingest Time & File Name

```python
additionalColumns = readJsonFiles.withColumn('ingestTime',current_timestamp()) \
                                .withColumn('fileName',col('_metadata.file_name'))
```

# Writing all data to delta

```python
additionalColumns.write.format('parquet') \
.mode('append') \
.save('/Volumes/ipl/bronze/ipldata')
```
